## 5. Data Quality Analysis




**5.1 Missing values identified and handled**

```python
# Missing values

missing = df.isnull().sum().sort_values(ascending=False)
missing_percentage = (df.isnull().mean() * 100).sort_values(ascending=False)

missing_summary = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": missing_percentage.round(2)
})

missing_summary[missing_summary["Missing_Count"] > 0]
```

```python
# Visualize missing values

plt.figure(figsize=(12, 5))
missing_plot = missing[missing > 0]
if len(missing_plot) > 0:
    sns.barplot(x=missing_plot.index, y=missing_plot.values)
    plt.xticks(rotation=60, ha="right")
    plt.ylabel("Number of Missing Values")
    plt.xlabel("Column")
    plt.title("Missing Values by Column")
    plt.tight_layout()
    plt.show()
else:
    print("No missing values found.")
```

### Missing Value Treatment

Choose an appropriate strategy based on the nature of each variable.

Possible approaches:
- Remove records only when justified.
- Fill numerical values using an appropriate statistic when appropriate.
- Fill categorical values using a suitable category or method.
- Clearly document why the chosen approach is reasonable.

Do not blindly replace every missing value with zero.


```python
# Example: median imputation for numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

print("Remaining missing values:", df.isnull().sum().sum())
```

**5.2 Duplicate records identified and handled**

```python
# Duplicate records

duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

```python
# Remove duplicate rows

df = df.drop_duplicates().reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
```

## 8. Outlier Analysis

Investigate potential outliers using statistical and visual methods.




```python
# Visual Comparison of Extreme Outliers

# Outlier Analysis: Visualizing key skewed metrics
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.boxplot(ax=axes[0, 0], data=df, y="Yield_Tonnes_Ha", color="#4c72b0")
axes[0, 0].set_title("Outliers in Yield (tonnes/ha)")

sns.boxplot(ax=axes[0, 1], data=df, y="Profit_INR", color="#55a868")
axes[0, 1].set_title("Outliers in Farm Profit (INR)")

sns.boxplot(ax=axes[1, 0], data=df, y="Production_Tonnes", color="#c44e52")
axes[1, 0].set_title("Outliers in Total Production (tonnes)")

sns.boxplot(ax=axes[1, 1], data=df, y="Water_Efficiency_t_per_1000m3", color="#8172b3")
axes[1, 1].set_title("Outliers in Water Efficiency")

plt.tight_layout()
plt.show()
```

```python
# IQR-based outlier summary for numerical columns

outlier_summary = []

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        "Column": col,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": count,
        "Outlier_Percentage": round(count / len(df) * 100, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary).sort_values(
    "Outlier_Count", ascending=False
)

outlier_summary
```

## 14. Key Insights
**1. Seasonal Record Volume Distribution**

>**Observation:** The dataset is unevenly distributed across seasons, with Kharif containing 1,779 records 44.5%, Rabi containing 1,627 records 40.7%, and Zaid containing only 594 records 14.8%.

>**Evidence:** Univariate count plot and seasonal summary table under Section 8 and Section 12.

>**Interpretation:** Agricultural records are heavily concentrated in the Kharif and Rabi cycles, while Zaid represents a much smaller portion of recorded farm periods.

>**Limitation:** This reflects sample distribution in the dataset rather than total national agricultural output or land area proportions.

**2. Yield Disparities by Crop Type**

>**Observation:** Sugarcane consistently yields significantly higher per hectare averaging over 38 to 53 tonnes/ha compared to all other crops like wheat, rice, or maize which hover between 1 to 3 tonnes/ha.

>**Evidence:** Crop and season summary tables in Section 12.

>**Interpretation:** Sugarcane is a massive biomass crop that skews per-hectare yield metrics compared to grain or pulse crops.

>**Limitation:** High yield in weight does not automatically equate to higher net financial profit due to vastly different cultivation costs and market prices.

**3. Negative Average Profits Across Most Seasons**

>**Observation:** The mean profit for Rabi and Zaid seasons is negative, and even in Kharif, many individual crop categories show negative mean profit values despite positive total revenues.

>**Evidence:** Statistical summary and seasonal economic aggregations in Section 7 and Section 12.

>**Interpretation:** High total production costs frequently outpace market revenues for multiple farm subsets in the dataset.

>**Limitation:** This shows financial association within recorded sample constraints and depends heavily on the accuracy of recorded cost and price columns.

**4. Rainfall Volume Variations Across Seasons**

>**Observation:** Rainfall levels peak sharply during the Kharif season averaging 849 mm, drop significantly during Rabi 437 mm, and reach their lowest points in Zaid 304 mm.

>**Evidence:** Descriptive statistical breakdown grouped by season in Section 7.

>**Interpretation:** Environmental water availability is heavily tied to the seasonal calendar, forcing different irrigation strategies.

>**Limitation:** Rainfall is aggregated at the record level and does not account for localized micro-climates or flash weather events.

**5. Irrigation Method Impact on Yield**

>**Observation:** Farms utilizing Drip irrigation show distinct yield and efficiency patterns compared to traditional Flood irrigation methods.

>**Evidence:** Bivariate analysis comparing irrigation methods and yield in Section 10.

>**Interpretation:** Modern irrigation techniques correlate with targeted resource delivery.

>**Limitation:** The data shows an association between irrigation method and yield, but does not prove that switching irrigation methods alone causes a direct increase in crop output without controlling for farm size and crop type.

**6. Farm Area and Production Distribution**

>**Observation:** Production volume scales up alongside farm area, but the distribution contains wide variance with a few large farms producing disproportionately high total tonnes.

>**Evidence:** Bivariate scatter plot of Farm Area vs. Production in Section 10.

>**Interpretation:** Larger land holdings naturally support higher absolute output volume.

>**Limitation:** Land size alone does not dictate efficiency, as smaller farms often achieve higher per-hectare yields through intensive management.

**7. Water Efficiency Disparities**

>**Observation:** Water efficiency fluctuates widely across different crops, with certain crops optimizing water usage much better than others during specific seasonal cycles.

>**Evidence:** Student Analysis 2 grouped bar chart and summary tables in Section 13.

>**Interpretation:** Certain crops yield higher tonnage per unit of water consumed, highlighting resource utilization differences.

>**Limitation:** Water efficiency calculations depend entirely on estimated water usage records rather than metered irrigation data.

**8. Disease and Pest Risk Distribution**

>**Observation:** Disease and pest risk percentages span widely across records ranging from 5% to nearly 89%, with intermediate averages resting around 46%.

>**Evidence:** Outlier summary and univariate distribution metrics in Section 7 and Section 9.

>**Interpretation:** Crop vulnerability varies substantially across different environmental conditions and farm practices.

>**Limitation:** Risk percentages represent modeled or recorded metric estimates rather than confirmed crop destruction events.



## 15. Recommendations



**Prioritize Cost Auditing and Expense Optimization for Grains:**

>**Finding:** Grains such as Rice, Wheat, and Maize show negative average profits across all three seasons  ,negative Rice average is ₹64,903 in Kharif and drops further in Zaid, despite producing consistent yields.

>**Recommendation:** Focus on controlling input costs specifically optimizing fertilizer and pesticide quantities rather than only attempting to increase volume output, as high expenditure directly creates financial losses for these staple crops.

>**Limitation:** Farm-level breakdown of individual expense categories labor, machinery, transport is unavailable, so specific cost-cutting targets must be verified on-site.

**Re-evaluate Crop Selection for the Zaid Season:**

>**Finding:** The Zaid season records an overall negative mean profit of ₹24,804 and has the highest average water usage 6,419 m³ relative to its lower aggregate production.

>**Recommendation:** Farmers should exercise caution when planting water-intensive crops during Zaid, shifting preference toward high-value, heat-tolerant options like Chilli or specific pulses only when efficient irrigation is accessible.

>**Limitation:** Seasonal profitability varies by region and local micro-climate, so crop selection cannot follow a one-size-fits-all policy without local market testing.

**Promote Micro-Irrigation Drip or Sprinkler Over Flood Irrigation:**

>**Finding:** Water efficiency varies widely across crops and seasons, while flood irrigation accounts for the highest share of records without providing a proportionate increase in yield.

>**Recommendation:** Transitioning toward targeted micro-irrigation methods like Drip can conserve water resources while maintaining steady yields, especially during dry Rabi and Zaid periods.

>**Limitation:** Capital costs and initial equipment setup expenses for micro-irrigation systems are not captured in the dataset, which may limit feasibility for smallholder farms.

**Implement Season-Specific Integrated Pest Management :**

>**Finding:** Disease and pest risk scores average significantly higher in Kharif 54.5% compared to Rabi 40.5% and Zaid 38.2%, tracking high humidity and rainfall spikes.

>**Recommendation:** Deploy preventive pest management interventions early in the Kharif cycle rather than waiting for outbreak conditions, keeping monitoring proactive during high-humidity weeks.

>**Limitation:** The dataset records a risk score percentage rather than specific pathogen types or recorded crop failure events, so exact chemical or biological controls must be selected per crop.


## 16. Conclusion

**Dataset Overview & Scope:**

The analysis evaluated 4,000 farm records across 28 features, examining environmental conditions, farming practices, input resource volumes, production yields, and financial returns across Kharif, Rabi, and Zaid seasons. Missing values across rainfall, soil moisture, and yield were systematically imputed using median values to preserve data structure without introducing artificial skew.

**Core Patterns Discovered:**

>**Seasonal Imbalance:** Agricultural activity in the dataset is heavily centered on Kharif 44.5% and Rabi 40.7%, with Zaid representing a limited footprint 14.8%.

>**Biomass vs. Economic Performance:** High yield does not guarantee high profitability. Sugarcane produces the highest biomass yields 38 to 53 tonnes/ha and high profits, but commercial cash crops like Chilli achieve solid profits even with modest yields. Conversely, standard grains Wheat, Rice, Maize struggle with negative average margins due to elevated total operational costs.

>**Resource & Climate Trends:** Rainfall and humidity drop substantially from Kharif to Zaid, requiring greater reliance on artificial irrigation during warmer periods.

**Key Analytical Limitations:**

* The dataset provides cross-sectional records without multi-year historical tracking, limiting the ability to assess year-over-year climate trends.

* Relationships observed reflect correlation rather than direct causation.

* High-level financial figures lack itemized breakdown for labor, land lease, fuel, and storage costs.

**Future Decision-Making Value:**

This analysis provides a structured baseline for regional farm planners and extension officers to identify high-risk seasons, optimize irrigation methods, and assist farmers in selecting crop combinations that balance water consumption with positive net margins.
